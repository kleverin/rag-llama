## Technical Implementation

The system is built as a **Retrieval-Augmented Generation (RAG)** pipeline that ingests structured manufacturing data and audio sensor recordings, stores them in a vector database, and exposes a conversational interface over a local network.

---

### System Architecture

```
┌─────────────────────────────────────────────────────────────┐
│                        DATA SOURCES                         │
│                                                             │
│  119 × Shift Report (.xlsx)    MB4000 Sensor (.wav files)  │
│  MTConnect Operation Log       (timestamped recordings)     │
└────────────────┬───────────────────────┬────────────────────┘
                 │                       │
                 ▼                       ▼
┌───────────────────────┐   ┌───────────────────────────────┐
│   Excel Ingestion     │   │     Audio Processing          │
│   (ingest.py)         │   │     (audio_processing.py)     │
│                       │   │                               │
│ • pd.read_excel()     │   │  Stage 1: Load + Normalize    │
│ • Each row → text doc │   │  Stage 2: Feature Extraction  │
│   with sheet/row meta │   │   - RMS energy                │
│                       │   │   - Kurtosis (spikiness)      │
│                       │   │   - Crest factor              │
│                       │   │   - Dominant frequency (Hz)   │
│                       │   │   - Low-band energy ratio     │
│                       │   │   - Fault flag (rule-based)   │
│                       │   │  Stage 3: Serialize → text    │
└──────────┬────────────┘   └──────────────┬────────────────┘
           │                               │
           └──────────────┬────────────────┘
                          ▼
┌─────────────────────────────────────────────────────────────┐
│                     EMBEDDING & INDEXING                    │
│                                                             │
│   nomic-embed-text (via Ollama)  →  ChromaDB vector store  │
│   Each document chunk embedded as a high-dimensional vector │
└─────────────────────────────────┬───────────────────────────┘
                                  │
                                  ▼
┌─────────────────────────────────────────────────────────────┐
│                    RAG QUERY ENGINE                         │
│                                                             │
│   LlamaIndex  •  chat_mode="condense_plus_context"         │
│   LLM: llama3.2 (Ollama, runs fully local)                 │
│   Retrieves top-5 most relevant chunks per query           │
└─────────────────────────────────┬───────────────────────────┘
                                  │
                                  ▼
┌─────────────────────────────────────────────────────────────┐
│                    FASTAPI SERVER (server.py)               │
│                                                             │
│   POST /ask    →  natural-language question → answer        │
│   GET  /       →  built-in chat UI (phone-accessible)      │
│   POST /reset  →  clear conversation history               │
│   GET  /health →  liveness check                           │
│                                                             │
│   Accessible from any device on the LAN (0.0.0.0:8000)    │
└─────────────────────────────────────────────────────────────┘
```

---

### Key Design Decisions

| Choice | Reason |
|--------|--------|
| **Fully local LLM (Ollama + llama3.2)** | No data leaves the network — important for proprietary manufacturing data |
| **ChromaDB persistent store** | Ingest once, query many times; survives server restarts |
| **Row-level Excel chunking** | Each shift report row becomes its own document so fine-grained date/part queries work |
| **Audio feature extraction → text** | Converts acoustic signals into text descriptors so they can be retrieved alongside tabular data by the same vector search |
| **Fault flag (kurtosis > 4 AND crest factor > 5)** | Rule-based early warning derived from established vibration analysis thresholds |
| **CORS open (`*`)** | Intentional — allows the chat UI to be used from a phone on the factory floor LAN |

## Evaluation and Testing

### Functional Testing of the RAG System

The system was tested by querying it with questions derived directly from the ingested shift report data. Two representative queries were run:

**Query 1:** *"How many parts were completed in May 2025?"*

The system successfully retrieved and answered for dates where data was correctly ingested:
- May 5, 2025 → **123 parts completed**
- May 29, 2025 → **156 parts completed**

Dates outside these records returned no data, which exposed a known gap: not all shift report rows were ingested in the initial run. Some files were skipped due to formatting inconsistencies in the source Excel files. This is a data completeness issue, not a model failure — the retrieval and generation pipeline functioned correctly for the data that was present.

**Query 2:** *"How many parts were a failure in 2025?"*

The system returned part numbers along with produced vs. completed counts and calculated failures for the records it had access to. For aggregate failure counts across the full year, the model correctly indicated it did not have sufficient data — confirming that the system does not hallucinate when data is missing.

This has since been incorporated into the development roadmap.

## Expected Outcome

### Current State of Operations

Currently, operators and engineers at the Kirby Risk automation cell rely on shift reports stored as individual Excel files spread across multiple directories. Answering a question like "what happened on a specific day" or "how many failures occurred last month" requires manually opening and searching through dozens of files — a process that is time-consuming and error-prone.

---

### Benefits After Implementation

Once fully deployed, the system will allow operators and engineers to query all historical data conversationally, including:

1. **On-demand reports** — Ask what happened on a specific day, week, month, or year and receive a consolidated summary without opening any files. The system supports natural language queries across the entire historical dataset.

2. **Failure analysis** — Query which parts failed, how many were produced vs. completed successfully, and retrieve the shift and operator information for that time window — enabling fast root cause investigation.

3. **Workforce traceability** — Cross-reference part failure events with who was on shift at that time, supporting accountability and targeted training decisions.

4. **Production scheduling** — Given a part number or part type, the model can retrieve historical machining records to estimate how long the job will take and suggest optimal scheduling windows based on past performance data.

5. **Audio-correlated fault detection** — By combining shift report data with acoustic sensor features (RMS, kurtosis, crest factor, fault flags), the system will be able to identify whether a logged part failure correlated with an abnormal acoustic event — giving engineers diagnostic context they currently lack.

---

### Primary Concern

The most critical dependency is **data completeness**. The system is only as useful as the data that has been ingested — gaps in shift reports or missing audio files lead to incomplete answers. Ensuring a clean, comprehensive ingestion of all historical records before handoff is the highest priority.

### Implementation Challenges

| Challenge | Description |
|-----------|-------------|
| **Data quality** | Source `.xlsx` files vary in formatting across reporting periods, requiring preprocessing before ingestion |
| **Audio processing volume** | Processing a large number of sensor `.wav` files is computationally intensive on local hardware |
| **Permanent adoption** | For long-term use, the server would need to run on a dedicated on-site machine with a defined process for ingesting new shift reports as they are generated |
| **User adoption** | Operators on the floor need minimal training to use the chat interface — the conversational UI is intentionally simple for this reason |
